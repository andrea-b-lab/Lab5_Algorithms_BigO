```python

# EXERCISE 3: E-COMMERCE ORDER PROCESSING

import random
import time

# 1. Student-specific inputs
last_name = input("Enter your surname: ").strip().upper()
student_id = input("Enter your student ID number: ").strip()
favorite_artist = input("Enter your favorite artist: ").strip().upper()

# 2. Generate required values
seed_digit = int(student_id[-1])
id_sum = sum(int(digit) for digit in student_id if digit.isdigit())
name_length = len(last_name)
artist_length = len(favorite_artist)

artist_sum = sum(
    ord(char) - ord('A') + 1
    for char in favorite_artist
    if 'A' <= char <= 'Z'
)

print("\nSTUDENT VALUES")
print("SEED_DIGIT:", seed_digit)
print("ID_SUM:", id_sum)
print("NAME_LENGTH:", name_length)
print("ARTIST_LENGTH:", artist_length)
print("ARTIST_SUM:", artist_sum)

# 3. Generate the original order dataset
n = 20
orders = [
    1000 + id_sum + artist_sum + i * 5
    for i in range(n)
]

# Create three conditions from the same orders
sorted_orders = sorted(orders)
reverse_orders = sorted(orders, reverse=True)

random.seed(seed_digit + id_sum + artist_sum)
random_orders = orders.copy()
random.shuffle(random_orders)

print("\nOriginal orders:", orders)
print("Sorted orders:", sorted_orders)
print("Reverse orders:", reverse_orders)
print("Random orders:", random_orders)


# 4. Bubble Sort
def bubble_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0
    n = len(arr)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            comparisons += 1

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 2
                swapped = True

        if not swapped:
            break

    return arr, comparisons, movements


# 5. Selection Sort
def selection_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0
    n = len(arr)

    for i in range(n - 1):
        min_index = i

        for j in range(i + 1, n):
            comparisons += 1

            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            movements += 2

    return arr, comparisons, movements


# 6. Insertion Sort
def insertion_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break

        arr[j + 1] = key
        movements += 1

    return arr, comparisons, movements


# 7. Test all algorithms on all input conditions
algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

conditions = {
    "Sorted": sorted_orders,
    "Reverse": reverse_orders,
    "Random": random_orders
}

results = []

for condition_name, dataset in conditions.items():
    print("\n" + "=" * 50)
    print("INPUT CONDITION:", condition_name)
    print("Input:", dataset)

    for algorithm_name, algorithm in algorithms.items():
        start = time.perf_counter()
        result, comparisons, movements = algorithm(dataset)
        elapsed = time.perf_counter() - start

        verified = result == sorted(dataset)

        print("\nAlgorithm:", algorithm_name)
        print("Sorted result:", result)
        print("Comparisons:", comparisons)
        print("Movements:", movements)
        print("Execution time:", elapsed, "seconds")
        print("Sorting verified:", verified)

        results.append({
            "Condition": condition_name,
            "Algorithm": algorithm_name,
            "Comparisons": comparisons,
            "Movements": movements,
            "Time (seconds)": elapsed,
            "Verified": verified
        })


# 8. Summary table
print("\nFINAL PERFORMANCE SUMMARY")
print(
    f"{'Condition':<10} {'Algorithm':<16} "
    f"{'Comparisons':>12} {'Movements':>10} {'Time (s)':>12}"
)

for row in results:
    print(
        f"{row['Condition']:<10} "
        f"{row['Algorithm']:<16} "
        f"{row['Comparisons']:>12} "
        f"{row['Movements']:>10} "
        f"{row['Time (seconds)']:>12.8f}"
    )


# 9. Big O analysis
print("\nBIG O ANALYSIS")
print("Bubble Sort: Best O(n), Average O(n^2), Worst O(n^2)")
print("Selection Sort: Best/Average/Worst O(n^2)")
print("Insertion Sort: Best O(n), Average O(n^2), Worst O(n^2)")
print("Auxiliary space: O(1) for each sorting algorithm")
```

```python

# EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS

# 1. Student-specific inputs
last_name = input("Enter your surname: ").strip().upper()
student_id = input("Enter your student ID number: ").strip()
favorite_artist = input("Enter your favorite artist: ").strip().upper()

# 2. Generate required student-specific values
seed_digit = int(student_id[-1])
id_sum = sum(int(digit) for digit in student_id if digit.isdigit())
name_length = len(last_name)
artist_length = len(favorite_artist)

artist_sum = sum(
    ord(char) - ord('A') + 1
    for char in favorite_artist
    if 'A' <= char <= 'Z'
)

print("\nSTUDENT VALUES")
print("SEED_DIGIT:", seed_digit)
print("ID_SUM:", id_sum)
print("NAME_LENGTH:", name_length)
print("ARTIST_LENGTH:", artist_length)
print("ARTIST_SUM:", artist_sum)

# 3. Generate two ascending datasets
n = 10

department_a = [
    seed_digit + id_sum + i * 3
    for i in range(n)
]

department_b = [
    artist_length + name_length + i * 4
    for i in range(n)
]

print("\nDepartment A dataset:", department_a)
print("Department B dataset:", department_b)

# 4. Merge the two sorted datasets
def merge_sorted_lists(a, b):
    i = 0
    j = 0
    merged = []
    comparisons = 0

    while i < len(a) and j < len(b):
        comparisons += 1

        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1

    # Copy any remaining values
    merged.extend(a[i:])
    merged.extend(b[j:])

    return merged, comparisons


# 5. Execute merge
merged_dataset, comparisons = merge_sorted_lists(
    department_a, department_b
)

# 6. Verify the result
expected_count = len(department_a) + len(department_b)

print("\nCONSOLIDATED DATASET:", merged_dataset)
print("Records before consolidation:", expected_count)
print("Records after consolidation:", len(merged_dataset))
print("Comparison count:", comparisons)
print("Correct number of records:", len(merged_dataset) == expected_count)
print("Sorted order verified:", merged_dataset == sorted(merged_dataset))

# 7. Complexity analysis
print("\nCOMPLEXITY ANALYSIS")
print("Time complexity: O(n + m)")
print("Auxiliary space: O(n + m)")
```

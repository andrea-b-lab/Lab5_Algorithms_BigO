```python
# [CELL 3] LINEAR GROWTH
def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in [10,100,1000,5000,10000]:
    print(f"n={n:5d} | Operations={linear_operations(n):5d}")
```

```python
# [CELL 4] BIG O CLASSIFICATION
def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in [10,100,1000]:
    print(
        f"n={n:4d} | O(1)=1 | "
        f"O(log n)={log_ops(n):2d} | "
        f"O(n)={n:4d} | O(n^2)={n*n:7d} "
    )
```

```python
# [CELL 5] SPACE COMPLEXITY
def constant_space(n):
    total = 0 
    for value in range(n):
        total += value
    return total
def linear_space(n):
    return list(range(n))
for n in [10,100,1000, 5000]:
    data = linear_space(n)
    print(f"n={n:5d} | O(1): constant | O(n): {len(data)}")
```

```python
# [CELL 6] CASE ANALYSIS
def search_case(data, target):
    for i, value in enumerate(data, 1):
        if value == target:
            return i
    return len(data)
data = [11, 22, 33, 44, 55]

for case, target in {"Best": 11, "Average": 33, "Worst": 99}.items():
    print(f"{case} Case:{search_case(data, target)} comparisons")
```

```python
# [CELL 7] LINEAR SEARCH
def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
print("Dataset Size:", len(data))
```

```python
# [CELL 8] BINARY SEARCH
def binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons
index, comparisons = binary_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
print("Dataset Size:", len(data))
```

```python
# [CELL 9] SEARCH COMPARISON
data = list(range(1, 101))
target = ((ID_SUM + NAME_LENGTH) % 100) + 1

li, lc = linear_search(data, target)
bi, bc = binary_search(data, target)

print("Target:", target)
print(f"Linear Search: Index={li}, Comparisons={lc}")
print(f"Binary Search: Index={bi}, Comparisons={bc}")
print("Reduction:", lc - bc)
```

```python
# [CELL 10] BUBBLE SORT
import time

def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    movements = 0

    start_time = time.perf_counter()

    for i in range(len(a)):
        swapped = False

        for j in range(len(a) - i - 1):
            comparisons += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                movements += 2
                swapped = True

        if not swapped:
            break

    execution_time = time.perf_counter() - start_time

    return a, comparisons, swaps, movements, execution_time


data = list(range(1, 21))

data = data + [
    (ID_SUM * 3) % 100,
    (SEED_DIGIT * 11) % 100,
    (NAME_LENGTH * 7) % 100,
    42, 17, 89, 31, 56
]

sorted_data, comparisons, swaps, movements, execution_time = bubble_sort(data)

print("Dataset Size:", len(data))
print("Sorted Data:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
print("Movements:", movements)
print(f"Execution Time: {execution_time:.8f} seconds")
```

```python
# [CELL 11] SELECTION SORT
import time

def selection_sort(data):
    a = data.copy()
    comparisons = 0
    swaps = 0
    movements = 0

    start_time = time.perf_counter()

    for i in range(len(a) - 1):
        minimum = i

        for j in range(i + 1, len(a)):
            comparisons += 1

            if a[j] < a[minimum]:
                minimum = j

        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
            movements += 2

    execution_time = time.perf_counter() - start_time

    return a, comparisons, swaps, movements, execution_time


sorted_data, comparisons, swaps, movements, execution_time = selection_sort(data)

print("Dataset Size:", len(data))
print("Sorted Data:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
print("Movements:", movements)
print(f"Execution Time: {execution_time:.8f} seconds")
```

```python
# [CELL 13] INSERTION SORT
import time

def insertion_sort(data):
    a = data.copy()
    comparisons = 0
    shifts = 0
    movements = 0

    start_time = time.perf_counter()

    for i in range(1, len(a)):
        key, j = a[i], i - 1
        movements += 1  # Assignment ng key

        while j >= 0:
            comparisons += 1

            if a[j] <= key:
                break

            a[j + 1] = a[j]
            shifts += 1
            movements += 1
            j -= 1

        a[j + 1] = key
        movements += 1

    execution_time = time.perf_counter() - start_time

    return a, comparisons, shifts, movements, execution_time


sorted_data, comparisons, shifts, movements, execution_time = insertion_sort(data)

print("Dataset Size:", len(data))
print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
print("Movements:", movements)
print(f"Execution Time: {execution_time:.8f} seconds")
```

```python
# [CELL 13] EXECUTION-TIME COMPARISON
import time

def measure(sort_function, data, trials=5):
    times = []

    for _ in range(trials):
        start_time = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start_time)

    return result, sum(times) / trials


data_size = 100 + ID_SUM * 10

performance_data = [
    (ID_SUM * 3 + i * 7) % 1000
    for i in range(data_size)
]

bubble_result, bubble_time = measure(bubble_sort, performance_data)
selection_result, selection_time = measure(selection_sort, performance_data)
insertion_result, insertion_time = measure(insertion_sort, performance_data)

print(f"Dataset Size: {data_size}")
print(f"Bubble Sort: {bubble_time:.8f} seconds")
print(f"Selection Sort: {selection_time:.8f} seconds")
print(f"Insertion Sort: {insertion_time:.8f} seconds")
```

```python
# [CELL 14] SORTING PERFORMANCE COMPARISON
results = [
    ("Bubble Sort", bubble_result, bubble_time),
    ("Selection Sort", selection_result, selection_time),
    ("Insertion Sort", insertion_result, insertion_time)
]

print(f"{'Algorithm':<18} {'Comparisons':<12} {'Movements':>12} {'Time (s)':>12}")

for name, result, elapsed in results:
    print(
        f"{name:<18}{result[1]:>12}"
        f"{result[2]:>12}{elapsed:>12.8f}"
    )
```

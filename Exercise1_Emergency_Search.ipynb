```python
# EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH

import time

# 1. Student-specific inputs
last_name = input("Enter your surname: ").strip().upper()
student_id = input("Enter your student ID number: ").strip()
favorite_artist = input("Enter your favorite artist: ").strip().upper()

# 2. Generate student-specific values
seed_digit = int(student_id[-1])
id_sum = sum(int(digit) for digit in student_id if digit.isdigit())
name_length = len(last_name)
artist_length = len(favorite_artist)
artist_sum = sum(
    ord(char) - ord('A') + 1
    for char in favorite_artist
    if 'A' <= char <= 'Z'
)

print("\nSTUDENT VALUES")
print("Surname:", last_name)
print("Student ID:", student_id)
print("Favorite artist:", favorite_artist)
print("SEED_DIGIT:", seed_digit)
print("ID_SUM:", id_sum)
print("NAME_LENGTH:", name_length)
print("ARTIST_LENGTH:", artist_length)
print("ARTIST_SUM:", artist_sum)

# 3. Generate the original sorted incident records
n = 15
original = [
    1000 + (id_sum * 10) + (i * 10) + seed_digit
    for i in range(n)
]

# 4. Rotate the list using SEED_DIGIT
rotation = seed_digit % n
rotated = original[rotation:] + original[:rotation]

# 5. Generate the target incident using ID_SUM and ARTIST_SUM
target_index = (id_sum + artist_sum) % n
target = original[target_index]

print("\nOriginal dataset:", original)
print("Rotation point:", rotation)
print("Rotated dataset:", rotated)
print("Target incident:", target)


# 6. Binary Search for a rotated sorted list
def binary_search_rotated(arr, target):
    left = 0
    right = len(arr) - 1
    comparisons = 0

    while left <= right:
        mid = (left + right) // 2
        comparisons += 1

        if arr[mid] == target:
            return mid, comparisons

        # Determine which half is sorted
        if arr[left] <= arr[mid]:
            comparisons += 1
            if arr[left] <= target < arr[mid]:
                right = mid - 1
            else:
                left = mid + 1
        else:
            comparisons += 1
            if arr[mid] < target <= arr[right]:
                left = mid + 1
            else:
                right = mid - 1

    return -1, comparisons


# 7. Linear Search
def linear_search(arr, target):
    comparisons = 0

    for index, value in enumerate(arr):
        comparisons += 1
        if value == target:
            return index, comparisons

    return -1, comparisons


# 8. Execute and measure both searches
start = time.perf_counter()
binary_position, binary_comparisons = binary_search_rotated(
    rotated, target
)
binary_time = time.perf_counter() - start

start = time.perf_counter()
linear_position, linear_comparisons = linear_search(
    rotated, target
)
linear_time = time.perf_counter() - start


# 9. Display results
print("\nBINARY SEARCH RESULTS")
if binary_position != -1:
    print("Target found at index:", binary_position)
else:
    print("Target incident not found.")
print("Comparisons:", binary_comparisons)
print("Execution time:", binary_time, "seconds")

print("\nLINEAR SEARCH RESULTS")
if linear_position != -1:
    print("Target found at index:", linear_position)
else:
    print("Target incident not found.")
print("Comparisons:", linear_comparisons)
print("Execution time:", linear_time, "seconds")

print("\nCOMPLEXITY ANALYSIS")
print("Binary Search: O(log n) time, O(1) auxiliary space")
print("Linear Search: O(n) time, O(1) auxiliary space")
```
